```python
img = cv2.imread('chrono_image.png')
B, R, G = cv2.split(img)

#subplot do azul (B)
plt.subplot(1,3,1)
plt.imshow(B, cmap='grey')
plt.axis('off')
plt.title('Azul (B)')

#subplot do verde (G)
plt.subplot(1,3,2)
plt.imshow(G, cmap='grey')
plt.axis('off')
plt.title('Verde (G)')

#subplot do vermelho (R)
plt.subplot(1,3,3)
plt.imshow(R, cmap='grey')
plt.axis('off')
plt.title('Vermelho (R)')
```
