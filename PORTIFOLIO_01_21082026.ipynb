```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python
arq = files.upload()

img_bgr = cv2.imread('girotto.png')
img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
```

```python
def quantizar(img, bits):
    shift = 8 - bits
    return (img >> shift) << shift

img_1bit = quantizar(img_gray, 1)
img_2bit = quantizar(img_gray, 2)
img_4bit = quantizar(img_gray, 4)
img_8bit = quantizar(img_gray, 8)
```

```python
fig, ax = plt.subplots(1, 4, figsize=(20,5))

ax[0].imshow(img_1bit, cmap='gray', vmin=0, vmax=255)
ax[0].set_title('1 bit (2 níveis)')
ax[0].axis('off')

ax[1].imshow(img_2bit, cmap='gray', vmin=0, vmax=255)
ax[1].set_title('2 bits (4 níveis)')
ax[1].axis('off')

ax[2].imshow(img_4bit, cmap='gray', vmin=0, vmax=255)
ax[2].set_title('4 bits (16 níveis)')
ax[2].axis('off')

ax[3].imshow(img_gray, cmap='gray', vmin=0, vmax=255)
ax[3].set_title('8 bits (256 níveis)')
ax[3].axis('off')

plt.tight_layout()
plt.savefig('quantizacao.png')
plt.show()
```

## 1 bit (2 níveis):
  Praticamente em preto em branco,há apenas uma silhueta dos olhos, boca e gravata. Os detalhes do fundo sumiram completamente em preto.
## 2 bit (4 níveis):
  Já dá pra reconhecer melhor o rosto, mas com manchas cinzas claras, transições, principalmente no rosto. O fundo já aparece, mas com desenhos pretos.
## 4 bit (16 níveis):
  Agora a imagem já se parece muito mais próxima da original, as trasições são levemente perceptíveis, principalmente no fundo e menos no rosto.
## 8 bit (256 níveis):
  Exaramente como a original, mas em tons de cinza.
