```python
img_bgr = cv2.imread('girotto.jpg')
img_bgr
```

```python
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
img_rgb
```

```python
plt.figure(figsize=(5,3)) # quadro 500x300
plt.imshow(img_rgb)
plt.title('Girotto em RGB')
plt.axis('off')
plt.show()
```

```python
img_hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)
img_hsv
```

```python
fig, ax = plt.subplots(1, 3, figsize=(15,9))

ax[0].imshow(img_rgb)
ax[0].set_title('Girotto em RGB')
ax[0].axis("off")
ax[1].imshow(img_bgr)
ax[1].set_title('Girotto em BGR')
ax[1].axis("off")
ax[2].imshow(img_hsv)
ax[2].set_title('Girotto em HSV')
ax[2].axis("off")

plt.tight_layout()
plt.show()
```

```python
img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
img_gray
```

```python
fig, ax = plt.subplots(2, 2, figsize=(5,5))

ax[0][0].imshow(img_rgb)
ax[0][0].set_title('Girotto em RGB')
ax[0][0].axis("off")

ax[0][1].imshow(img_bgr)
ax[0][1].set_title('Girotto em BGR')
ax[0][1].axis("off")

ax[1][0].imshow(img_hsv)
ax[1][0].set_title('Girotto em HSV')
ax[1][0].axis("off")

ax[1][1].imshow(img_gray, cmap='gray')
ax[1][1].set_title('Girotto GRAY')
ax[1][1].axis("off")
ax[1][1]

plt.tight_layout()
plt.savefig('conversao_de_cores.png')
plt.show()
```
